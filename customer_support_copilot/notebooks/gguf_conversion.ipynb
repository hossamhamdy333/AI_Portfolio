```python
#  download + convert the LoRA adapter to GGUF separately
from huggingface_hub import snapshot_download

snapshot_download("hossam3759180/support-copilot-llama3-lora", local_dir="lora_adapter")
!python llama.cpp/convert_lora_to_gguf.py lora_adapter --base base_model --outfile lora.gguf
```

```python
!cd llama.cpp && cmake -B build
!cmake --build llama.cpp/build --config Release -j 2
!./llama.cpp/build/bin/llama-export-lora -m base.gguf -o support-copilot.gguf --lora lora.gguf
```

```python
#quantize for fast CPU inference
!./llama.cpp/build/bin/llama-quantize support-copilot.gguf support-copilot-q4.gguf Q4_K_M
```

```python
#push the quantized GGUF file to Hugging Face Hub
from huggingface_hub import HfApi, login
import getpass

login(token=getpass.getpass("HF token (write access): "))
api = HfApi()
api.create_repo("hossam3759180/support-copilot-gguf", exist_ok=True)
api.upload_file(
    path_or_fileobj="support-copilot-q4.gguf",
    path_in_repo="support-copilot-q4.gguf",
    repo_id="hossam3759180/support-copilot-gguf",
)
print("Uploaded to https://huggingface.co/hossam3759180/support-copilot-gguf")
```
